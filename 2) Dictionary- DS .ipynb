```python
my_dict = {"Name": "Aanya", "Roll_No": 34, "Marks": 90}

print("Keys:")

for key in my_dict.keys():
    print(key)

print("\nValues:")

for value in my_dict.values():
    print(value)

print("\nKeys and Values:")

for key, value in my_dict.items():
    print(key, ":", value)
```

```python
my_dict = {}

for i in range(1, 16):
    my_dict[i] = i ** 2

print(my_dict)
```

```python
my_dict = {1: 10, 2: 20, 3: 30, 4: 40}

total = sum(my_dict.values())

print("Sum of values:", total)
```

```python
my_dict = {
    "name": "Aanya",
    "roll_no": 34,
    "panel": "B",
    "age": 18,
    "marks": 90
}

print("Original dictionary:")
print(my_dict)

# Add information
my_dict["city"] = "Pune"

print("\nAfter adding information:")
print(my_dict)

# Delete information
del my_dict["age"]

print("\nAfter deleting information:")
print(my_dict)
```

```python
list1 = ["name", "panel", "rollno"]
list2 = ["ABC", "B", 34]

my_dict = dict(zip(list1, list2))

print(my_dict)
```

```python
my_dict = {
    "name": "ABC",
    "panel": "B",
    "rollno": 34
}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("Keys list:", keys_list)
print("Values list:", values_list)
```

```python
my_dict = {
    "marks1": 23,
    "marks2": 123,
    "marks3": 43,
    "marks4": 13,
    "marks5": 39
}

total = sum(my_dict.values())
mean = total / len(my_dict)

print("Mean of values:", mean)
```
