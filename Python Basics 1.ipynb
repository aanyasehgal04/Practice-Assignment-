```python
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")

sub1 = float(input("Enter marks of Subject 1: "))
sub2 = float(input("Enter marks of Subject 2: "))
sub3 = float(input("Enter marks of Subject 3: "))

percentage = (sub1 + sub2 + sub3) / 3

print("\nStudent Name:", name)
print("Roll Number:", roll_no)
print("Percentage:", percentage)

if sub1 >= sub2 and sub1 >= sub3:
    print("Highest Marks: Subject 1")
elif sub2 >= sub1 and sub2 >= sub3:
    print("Highest Marks: Subject 2")
else:
    print("Highest Marks: Subject 3")

if sub1 <= sub2 and sub1 <= sub3:
    print("Lowest Marks: Subject 1")
elif sub2 <= sub1 and sub2 <= sub3:
    print("Lowest Marks: Subject 2")
else:
    print("Lowest Marks: Subject 3")
```

```python
num = int(input("Enter a number: "))

if num < 2:
    print("Not Prime")
else:
    prime = True
    
    for i in range(2, num):
        if num % i == 0:
            prime = False
            break
    
    if prime:
        print("Prime")
    else:
        print("Not Prime")
```

```python
num = int(input("Enter a number: "))
sum_digits = 0

while num > 0:
    digit = num % 10
    sum_digits = sum_digits + digit
    num = num // 10

print("Sum of digits:", sum_digits)
```

```python
num = int(input("Enter a number: "))
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse:", reverse)
```

```python
num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```
