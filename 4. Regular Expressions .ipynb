```python
import re

pattern = r"^[bh][aiu]t$"

strings = ["bat", "bit", "but", "hat", "hit", "hut"]

for string in strings:
    if re.match(pattern, string):
        print(string, "Matched")
    else:
        print(string, "Not Matched")
```

```python
import re

pattern = r"^[A-Za-z]+ [A-Za-z]+$"

string = input("Enter first and last name: ")

if re.match(pattern, string):
    print("Matched")
else:
    print("Not Matched")
```

```python
import re

pattern = r"^[A-Za-z]+, [A-Za-z]$"

string = input("Enter word and initial: ")

if re.match(pattern, string):
    print("Matched")
else:
    print("Not Matched")
```
